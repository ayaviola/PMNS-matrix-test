### **Neutrino flavor mixing matrix and Neutrino Oscillations**

Neutrinos are neutrally charged fermions with a vanishingly small mass. A neutrino's precise mass remains inconclusive per its inherent quantum complexity that physicists thus far struggle to decipher, rendering its absolute value immeasurable for the time being. 

Knowing that neutrinos are fermions, they are specifically members of the lepton group and come in 3 distinct flavour eigenstates: **electron-neutrino**, **muon-neutrino**, **tau-neutrino**. This essentially means that for every charged lepton exists a corresponding neutrino.

```python
Dimension = sqrt(sum([d^2 for d in q_dims])) ; "total quantum dimension:" , Dimension
```
